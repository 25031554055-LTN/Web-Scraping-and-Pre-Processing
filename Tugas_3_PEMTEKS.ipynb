## Web Scraping with API

```python
import requests
from bs4 import BeautifulSoup

titles = []

url = 'https://books.toscrape.com/catalogue/page-'

for page in range(1, 4):
    url_page = url + str(page) + '.html'
    response = requests.get(url_page)
    soup = BeautifulSoup(response.text, 'html.parser')

    books = soup.find_all('h3')
    for book in books:
        title_tag = book.find('a')
        titles.append(title_tag['title'])

    print(f"Halaman {page} selesai — total: {len(titles)}")

for i, t in enumerate(titles):
    print(f"{i+1}. {t}")
```

```python
# using API
import requests
import time

url_api = "https://openlibrary.org/search.json?q=harry+potter&page="
titles_api = []
authors_api = []

# Adding a User-Agent helps prevent servers from refusing connections
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36'
}

for page in range(1, 4):
    url_page = url_api + str(page)
    try:
        response = requests.get(url_page, headers=headers, timeout=10)
        response.raise_for_status() # Raise error for bad responses
        data = response.json()
        docs = data.get('docs', [])

        for doc in docs:
            titles_api.append(doc.get('title'))
            authors_api.append(doc.get('author_name', ['Unknown']))

        print(f"Halaman {page} selesai — total: {len(titles_api)}")

        # Sleep for 2 seconds to avoid aggressive scraping and rate-limiting
        time.sleep(2)

    except requests.exceptions.RequestException as e:
        print(f"Gagal mengambil halaman {page}: {e}")
        break

for i in range(len(titles_api)):
    print(f"{i+1}. {titles_api[i]} | {authors_api[i]}")
```

Remove URL, HTTP, and e-mail

```python
import re

def remove_rls(text):
    text = re.sub(r'https?\/\S+', '', str(text))
    text = re.sub(r'http\S+', '', str(text))
    text = re.sub(r'www\S+', '', str(text))
    text = re.sub(r'\S+@\S+', '', str(text))
    return text

hasil_rl = remove_rls(hasil_cf)
print(hasil_rl)
```

Remove Punctuations

```python
import string
punc = string.punctuation
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', punc))

hasil_rp = remove_punctuation(hasil_rl)
print(hasil_rp)
```

```python
def remove_punctuation_list(text):
    # List Emoticon
    text = re.sub(r'(:\)|:-\)|:D|:-D|;\)|;-\))', '', text)
    text = re.sub(r'(:\(|:-\()', '', text)
    text = re.sub(r'(:v|:-v)', '', text)

    # remove punctuation
    text = re.sub(r'[!"#$%&\'()*+,\-./:;<=>?@\[\]^_`{|}~]', ' ', text)
    text = re.sub(r'\s+', ' ', text).strip()

    return text
```
